```python
import os

def rename_files_with_extension(folder_path, common_name):
    files = [f for f in os.listdir(folder_path) if os.path.isfile(os.path.join(folder_path, f))]
    files.sort()  # Optional: Sort alphabetically

    for i, filename in enumerate(files, start=1):
        name, ext = os.path.splitext(filename)
        new_name = f"{common_name}{i}{ext}"  # Use common name + index + extension
        old_path = os.path.join(folder_path, filename)
        new_path = os.path.join(folder_path, new_name)
        os.rename(old_path, new_path)

    print("✅ Files renamed successfully.")

# 🔧 Example usage
folder = r"C:\Users\hp\Desktop\ANN Project\crowd_behavior_data\panic"
common_name = 'panic_'  # <-- Common name jo har file ke sath ho, jaise 'frame1', 'frame2', ...
rename_files_with_extension(folder, common_name)
```

```python
import cv2
import os

def extract_frames(video_path, output_folder, interval_sec, num_images):
    # Folder create karo agar nahi hai
    os.makedirs(output_folder, exist_ok=True)

    # Video load karo
    cap = cv2.VideoCapture(video_path)

    if not cap.isOpened():
        print("❌ Error opening video file.")
        return

    fps = cap.get(cv2.CAP_PROP_FPS)
    frame_interval = int(fps * interval_sec)
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

    print(f"🎞 FPS: {fps}, Frame interval: {frame_interval} frames")

    count = 0
    saved = 0

    while cap.isOpened() and saved < num_images:
        ret, frame = cap.read()
        if not ret:
            break

        if count % frame_interval == 0:
            filename = os.path.join(output_folder, f"frame_{saved+1}.jpg")
            cv2.imwrite(filename, frame)
            print(f"✅ Saved {filename}")
            saved += 1

        count += 1

    cap.release()
    print("✅ Done extracting frames.")

# 🧪 Example usage:
video_file = r"C:\Users\hp\Desktop\test video.mp4"    # <-- Video ka path
output_dir = r"C:\Users\hp\Desktop\ANN Project\test2"       # <-- Output folder
seconds_interval = 1                       # <-- Har kitne seconds baad frame lena hai
number_of_images = 17                           # <-- Total kitni images chahiyein

extract_frames(video_file, output_dir, seconds_interval, number_of_images)
```
